```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
url='/content/data-export (1) (1).csv'
ds=pd.read_csv(url)
```

```python
ds.describe()
```

```python
ds.head()
```

```python
ds.columns=ds.iloc[0]
ds=ds.drop(index=0).reset_index(drop=True)
ds.columns=['Session primary channel group','Date+hour','Users','Sessions','Engaged sessions','Average engagement time per session','Engaged sessions per user','Events per session','Engagement rate','Event count']
ds.head(2)
```

```python
ds.info()
```

```python
ds.describe()
```

```python
ds.isnull().sum()
```

```python
ds.shape
```

```python
ds['Date+hour']=pd.to_datetime(ds['Date+hour'] ,format='%Y%m%d%H')
```

```python
ds.info()
```

```python
nu=ds.columns.drop(['Session primary channel group','Date+hour'])
ds[nu]=ds[nu].apply(pd.to_numeric,errors='coerce')
```

```python
ds.head()
```

```python
ds.info()
```

```python
ds.describe()
```

```python
plt.figure(figsize=(12,5))
ds.groupby('Date+hour')[["Users","Sessions"]].sum().plot(ax=plt.gca())
```

```python
plt.figure(figsize=(12,5))
sns.barplot(x='Session primary channel group',y='Users',data=ds)
```

```python
ds.head()
```

```python
plt.figure(figsize=(12,5))
sns.barplot(x='Session primary channel group',y='Average engagement time per session',data=ds,palette='magma')
```

```python
plt.figure(figsize=(12,5))
sns.boxplot(x='Session primary channel group',y='Engaged sessions per user',data=ds,palette="magma")
plt.grid(True)
```
